```python
from langgraph.graph import StateGraph, START,END
from langchain_openai import ChatOpenAI
from typing import TypedDict
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
model = ChatOpenAI()
```

```python
class BlogState(TypedDict):
    title: str
    outline: str
    content: str
    rate: int
```

```python
def create_outline(state: BlogState) -> BlogState:
    
    # create title
    # call llm to create outline
    prompt = f"Create an outline for a blog post titled '{state['title']}'"
    outline = model.invoke(prompt).content

    # update state with outline
    state['outline'] = outline

    return state
```

```python
def create_blog(state: BlogState) -> BlogState:
    
    prompt = f"Write a blog post based on the title '{state['title']}' using the following outline:\n{state['outline']}"
    
    content = model.invoke(prompt).content
    
    state['content'] = content
    
    return state
```

```python
def rate_blog(state: BlogState) -> BlogState:
    
    prompt = f"Rate the following blog post on a scale of 1 to 10:\n{state['content']} on the basis of the outline:\n{state['outline']}"
    
    rating = model.invoke(prompt).content
    
    state['rate'] = rating
    
    return state
```

```python
graph = StateGraph(BlogState)

## nodes
graph.add_node('create_outline',create_outline)
graph.add_node('create_blog',create_blog)
graph.add_node('rate_blog',rate_blog)

## edges
graph.add_edge(START,'create_outline')
graph.add_edge('create_outline','create_blog')
graph.add_edge('create_blog','rate_blog')
graph.add_edge('rate_blog',END)

workflow = graph.compile()
```

```python
initial_state = BlogState(title="The Future of AI")

final_state = workflow.invoke(initial_state)

print(final_state['content'])
```
